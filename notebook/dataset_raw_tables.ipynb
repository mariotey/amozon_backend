# Raw Amazon review and item data

Download the Amazon Reviews 2023 source data using `datasets.load_dataset`, following the dataset settings in `config.py`. Run the cells from top to bottom with the project's Python environment and internet access.

This notebook only downloads the source datasets, exposes them as pandas tables, and previews them. It does not clean, rename, filter, aggregate, or engineer features. Source columns, text, identifiers, timestamps, and nested fields are retained.

There is no separate user table in the source configurations used by this project: user identifiers are stored in the review table's `user_id` column. The project's standalone user table is generated by aggregation in `data_transform.py`; it is intentionally not created here because that would require transformation.


```python
from pathlib import Path
import sys

# Support launching Jupyter from the repository root or its notebook folder.
project_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "config.py").is_file()
     and (path / "data_loader").is_dir()),
    None,
)
if project_root is None:
    raise RuntimeError("Launch Jupyter from this repository or its notebook folder.")
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from datasets import load_dataset
from IPython.display import display
from config import DATASET_NAME, DATASET_CATEGORY, DATASET_SPLIT_PERCENTAGE
```

## Raw reviews (including users)

Use the same configured review slice as `data_download.py` (currently the first 5%). Set `review_split = "full"` below if you want all reviews instead. No further row filtering is applied. `review_dataset` retains the downloaded dataset, and `review_df` provides a table view of all downloaded records.


```python
review_split = f"full[:{DATASET_SPLIT_PERCENTAGE}%]"

review_dataset = load_dataset(
    DATASET_NAME,
    f"raw_review_{DATASET_CATEGORY}",
    split=review_split,
)
review_df = review_dataset.to_pandas()
display(review_df.head())
```

## Raw item metadata

Download the full metadata split, matching `data_download.py`. `item_dataset` retains the downloaded dataset, and `item_df` contains all original metadata columns and records. Previews show only the first five rows without changing either table.


```python
item_dataset = load_dataset(
    DATASET_NAME,
    f"raw_meta_{DATASET_CATEGORY}",
    split="full",
)
item_df = item_dataset.to_pandas()
display(item_df.head())
```
